```python
from datetime import datetime
import traceback

try:
  x = 10 / 0
except Exception:
  log = traceback.format_exc()
  print(log)
  with open("error.log", "a", encoding="utf-8") as f:
    f.write(f"{datetime.now()}, 發生錯誤：\n{log}\n")
finally:
  print("程式結束")
```

```python
from datetime import datetime
import traceback

try:
  a = float(input("請輸入被除數："))
  b = float(input("請輸入除數："))
  result = a / b
  print("結果", result)
except ValueError:
  with open("log_test.txt", "a", encoding="utf-8") as f:
    f.write(f"{datetime.now()},發生輸入值錯誤\n")
    f.write(traceback.format_exc())
except ZeroDivisionError:
  print("錯誤：除數不能為零！")
  with open("log_test.txt", "a", encoding="utf-8") as f:
    f.write(f"{datetime.now()},發生除以零錯誤\n")
    f.write(traceback.format_exc())
finally:
  print("程式結束")
```
