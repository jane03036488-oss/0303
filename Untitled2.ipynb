```python
listA = []
while True:
    # 修正 1：將 inpue 改為 input
    item = input("請輸入A倉文具，輸入'/'結束：")
    if item == "/":
        break
    # 修正 2：將逗號 , 改為點 .
    listA.append(item)

listB = []
while True:
    # 修正 3：修正 input 拼字與縮排
    item = input("請輸入B倉文具，輸入'/'結束：")
    if item == "/":
        break
    # 修正 4：B 倉輸入的項目應該放到 listB，而不是 listA
    listB.append(item)

# 組合 A 倉與 B 倉
listC = listA + listB

# 修正 5：append 必須使用小括號 ()，並加入 "膠水"
listC.append("膠水")

# 將列表用 / 串接成字串
listC_str = "/".join(listC)

print(f"二倉合併清單：{listC_str}")
```
