```python
def c_bmi(h, w):
    bmi = w / (h / 100) ** 2
    return bmi


bmi1 = c_bmi(170, 60)
bmi2 = c_bmi(165, 55)
bmi3 = c_bmi(180, 80)


print(f"BMI 1: {bmi1:.2f}")
print(f"BMI 2: {bmi2:.2f}")
print(f"BMI 3: {bmi3:.2f}")
```
