```python
weapons = ["木棒", "手槍", "霰彈槍", "狙擊槍", "雷射劍"]
attacks = (10, 30, 50, 80, 100)

for i, name in enumerate(weapons, start=1):
    print(f"{i} {name}")

while True:
    n = int(input("請輸入武器編號（0 結束）："))

    if n == 0:
        break
    elif 1 <= n <= len(weapons):
        print(f"{weapons[n - 1]} 的攻擊力是 {attacks[n - 1]}")
    else:
        print("找不到武器")

avg = sum(attacks) / len(attacks)
print(f"平均攻擊力：{avg}")
```
