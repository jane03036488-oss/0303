```python
def calculate_bill(price, quantity):
    subtotal = price * quantity
    tax = subtotal * 0.05
    total = subtotal + tax
    return subtotal, tax, total


p = float(input("請輸入單價："))
q = int(input("請輸入數量："))

# 修正 1：將 add 改為正確的函數名稱 calculate_bill
x, y, z = calculate_bill(p, q)

# 修正 2：移除多餘的 } 符號
print(f"小計: {x}\n稅金: {y}\n總金額: {z}")
```

```python
def f(n):
    # 基本狀況：第 1 階有 1 種方法，第 2 階有 2 種方法
    if n == 1:
        return 1
    if n == 2:
        return 2

    # 遞迴呼叫：前兩階的方法數相加
    return f(n - 1) + f(n - 2)


# 呼叫函式並印出結果
result = f(10)
print(f"總共有{result}種方式")
```
