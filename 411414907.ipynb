```python
import math

r = float(input("請輸入半徑："))
c = 2 * math.pi * r
print(f"圓周長為 {c:.2f}")
```

```python
import math

# 請使用者輸入角度
degree = float(input("請輸入角度（度）："))

# 將角度轉換為弧度
a = degree * (math.pi / 180)
# 計算 sin 值
b = math.sin(a)

# 輸出結果（變數名稱修正為 b，並取小數點後兩位）
print(f"你計算的sin值為:{b:.2f}")
```
