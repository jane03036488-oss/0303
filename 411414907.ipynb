```python
def calculate_bill(price, quantity):
    subtotal = price * quantity
    tax = subtotal * 0.05
    total = subtotal + tax
    return subtotal, tax, total


p = float(input("請輸入單價："))
q = int(input("請輸入數量："))

# 修正 1：將 add 改為正確的函數名稱 calculate_bill
x, y, z = calculate_bill(p, q)

# 修正 2：移除多餘的 } 符號
print(f"小計: {x}\n稅金: {y}\n總金額: {z}")
```
