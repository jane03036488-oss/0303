```python
import math

r = float(input("請輸入半徑："))
c = 2 * math.pi * r
print(f"圓周長為 {c:.2f}")
```

```python
import math

# 請使用者輸入角度
degree = float(input("請輸入角度（度）："))

# 將角度轉換為弧度
a = degree * (math.pi / 180)
# 計算 sin 值
b = math.sin(a)

# 輸出結果（變數名稱修正為 b，並取小數點後兩位）
print(f"你計算的sin值為:{b:.2f}")
```

```python
s=7
guess=int(input("猜猜一個數:"))
while guess != s:
  if guess>s:
    print("你猜太大了")
  else:
    print("你猜太小了")
  guess=int(input("請再猜一個數:"))
print("恭喜中了")
```

```python
import random

s = random.randint(0, 9)
guess = int(input("請猜: "))

while guess != s:
    guess = int(input("答錯了，再猜: "))

print("恭喜你，猜對了！")
```

```python
import random

day = 1
max_day = 3
win = False

while day <= max_day and not win:
    lucky_number = random.randint(1, 50)
    guess = int(input(f"今天第{day}天,請輸入1-50數字"))
    print(f"今天的幸運數字是{lucky_number}")

    if guess == lucky_number:
        print("恭喜你中獎了！")
        win = True
    else:
        print("你沒中獎，明天再來\n")

    # 記得讓天數 +1，才能進入下一天
    day += 1

if not win:
    print("很遺憾3天都沒中獎，下次再試試吧！")
```
